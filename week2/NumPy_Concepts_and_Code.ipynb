# 1. What is NumPy, and Why Does it Matter for ML?

**NumPy** provides a fast, memory-efficient array type called `ndarray` (n-dimensional array), plus a huge collection of functions to operate on it.

### Why not just use Python lists?
- Python lists can hold *mixed* types and are stored as separate objects scattered in memory — flexible, but slow for numeric work.
- NumPy arrays store data of a **single type** in one contiguous memory block, and operations run as optimized, compiled C code under the hood.
- This means NumPy operations on large datasets can be **10x–100x faster** than equivalent pure-Python loops.

### Where NumPy fits in ML
- Pandas DataFrames are built on top of NumPy arrays.
- Scikit-learn expects your features (`X`) and labels (`y`) as NumPy arrays.
- TensorFlow and PyTorch tensors behave almost identically to NumPy arrays (and can convert to/from them easily).

In short: **NumPy is the numeric foundation everything else in this course is built on.**

<a id="creating-arrays"></a>
# 3. Creating Arrays

The simplest way to create an array is with `np.array()`, passing in a Python list (or nested lists for multiple dimensions).

```python
marks_list = [78, 85, 91, 67, 88]
marks_array = np.array(marks_list)

print(marks_array)
print(type(marks_array))
```

```python
# A 2D array from a nested list -- rows and columns, like a small data table
student_marks = np.array([
    [78, 85, 90],   # Ali
    [88, 92, 95],   # Sara
    [55, 62, 58],   # Ahmed
])

print(student_marks)
```

**Common mistake:** Mixing types inside the list passed to `np.array()` (e.g. `[1, 2, "three"]`). NumPy will "upcast" everything to a single common type (often turning numbers into strings), silently changing your data's meaning. Keep arrays single-typed, especially for numeric work.

<a id="attributes"></a>
# 4. Array Attributes

Every array carries useful metadata about its structure:

| Attribute | Meaning |
|---|---|
| `.shape` | tuple describing the size along each dimension |
| `.ndim` | number of dimensions |
| `.size` | total number of elements |
| `.dtype` | the data type of the elements |

```python
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
])

print("Shape:", student_marks.shape)   # (3 rows, 3 columns)
print("Dimensions:", student_marks.ndim)
print("Total elements:", student_marks.size)
print("Data type:", student_marks.dtype)
```

### ML Connection
`.shape` is one of the very first things you'll check when working with real datasets — it tells you how many samples (rows) and features (columns) you have, and mismatched shapes are one of the most common sources of errors when training ML models.

<a id="creation-functions"></a>
# 5. Array Creation Functions

NumPy provides shortcuts for creating arrays without manually typing every value — very useful for initializing weights, generating index ranges, or creating placeholder arrays.

```python
print(np.zeros(5))          # 1D array of five zeros
print(np.zeros((2, 3)))     # 2D array (2 rows, 3 columns) of zeros
print(np.ones((3, 2)))      # 2D array of ones
print(np.full((2, 2), 7))   # array filled with a constant value (7)
print(np.eye(3))            # 3x3 identity matrix (1s on the diagonal)
```

```python
print(np.arange(0, 10, 2))       # like range(): start, stop (excluded), step
print(np.linspace(0, 1, 5))     # 5 evenly spaced numbers from 0 to 1 (inclusive)
```

**Common mistake:** Confusing `np.arange(start, stop, step)` (step size) with `np.linspace(start, stop, num)` (number of points). `arange` may or may not land exactly on `stop`; `linspace` always includes both `start` and `stop` and controls the **count** of points instead.

<a id="random"></a>
# 6. Random Arrays

NumPy's random module is heavily used in ML for initializing weights, shuffling data, and creating synthetic datasets for practice. Setting a **seed** makes results reproducible — important for debugging and sharing experiments.

```python
rng = np.random.default_rng(seed=42)   # modern, recommended random generator

print(rng.random(5))              # 5 random floats in [0, 1)
print(rng.integers(1, 100, 5))    # 5 random integers in [1, 100)
print(rng.normal(0, 1, 5))        # 5 samples from a standard normal distribution
```

### ML Connection
Reproducibility matters enormously in ML: using a fixed seed means you (or a teammate) can re-run an experiment and get the exact same random split of data or the exact same initial weights.

<a id="indexing"></a>
# 7. Indexing and Slicing

1D array indexing/slicing works just like Python lists. 2D arrays use a comma to separate the row and column selectors: `array[row, column]`.

```python
marks = np.array([78, 85, 91, 67, 88])

print(marks[0])        # first element
print(marks[-1])       # last element
print(marks[1:4])      # slice: elements at index 1,2,3
```

```python
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
])

print(student_marks[0, 1])     # row 0, column 1 -> 85
print(student_marks[1])        # entire row 1 -> [88, 92, 95]
print(student_marks[:, 0])     # entire column 0 -> [78, 88, 55]
print(student_marks[0:2, 1:3]) # sub-grid: rows 0-1, columns 1-2
```

**Common mistake:** Writing `student_marks[1][2]` instead of `student_marks[1, 2]`. Both usually work and give the same result, but `[row, column]` is the idiomatic NumPy style and is significantly faster.

<a id="boolean-indexing"></a>
# 8. Boolean Indexing (Filtering)

You can filter an array using a condition, which produces a **boolean mask** — an array of `True`/`False` values the same shape as the original. Passing that mask back into `[]` returns only the elements where the mask is `True`.

```python
marks = np.array([78, 85, 91, 67, 88])

mask = marks >= 80
print(mask)              # [False  True  True False  True]
print(marks[mask])       # only the values >= 80

# Usually written in one line:
print(marks[marks >= 80])
```

### ML Connection
This is exactly the filtering pattern you'll use in Pandas (e.g. `df[df["age"] >= 18]`) — Pandas' boolean filtering is built directly on this NumPy mechanism.

<a id="fancy-indexing"></a>
# 9. Fancy Indexing

You can also index an array using a **list (or array) of specific indices** you want, in any order — this is called "fancy indexing."

```python
marks = np.array([78, 85, 91, 67, 88])

selected_indices = [0, 2, 4]
print(marks[selected_indices])   # elements at positions 0, 2, and 4

print(marks[[4, 0]])              # order matters -- returns in the order requested
```

<a id="reshaping"></a>
# 10. Reshaping Arrays

`.reshape()` changes an array's shape **without changing its data or total number of elements**. `.flatten()` (or `.ravel()`) collapses any array back into 1D.

```python
numbers = np.arange(1, 13)   # 1D array: [1, 2, ..., 12]
print(numbers.shape)

grid = numbers.reshape(3, 4)   # reshape into 3 rows, 4 columns
print(grid)
print(grid.shape)

flat_again = grid.flatten()
print(flat_again)
```

**Common mistake:** Trying to reshape into dimensions whose product doesn't match the original size — for example, reshaping 12 elements into `(3, 5)` raises a `ValueError`, since `3 * 5 = 15 ≠ 12`.

```python
numbers = np.arange(1, 13)

# numbers.reshape(3, 5)   # Uncomment to see the ValueError: cannot reshape array of size 12 into shape (3,5)

# A handy trick: use -1 to let NumPy figure out one dimension automatically
auto_reshaped = numbers.reshape(4, -1)   # 4 rows, columns computed automatically
print(auto_reshaped.shape)
```

<a id="arithmetic-broadcasting"></a>
# 11. Array Arithmetic and Broadcasting

Arithmetic on NumPy arrays is **element-wise** by default — no loops needed. **Broadcasting** is NumPy's rule for automatically applying an operation between arrays of *different but compatible* shapes (for example, an array and a single number).

```python
marks = np.array([78, 85, 91, 67, 88])

print(marks + 5)     # add 5 to every element
print(marks * 2)     # double every element
print(marks / 100)   # convert every mark to a proportion of 100
```

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)    # element-wise addition -- pairs up matching positions
print(a * b)    # element-wise multiplication
```

```python
# Broadcasting: a 2D array combined with a 1D array
grades = np.array([
    [78, 85, 90],
    [88, 92, 95],
])

bonus_per_subject = np.array([2, 0, 5])   # one value per column

print(grades + bonus_per_subject)   # bonus_per_subject is "stretched" across every row
```

**Common mistake:** Assuming `list1 + list2` on plain Python lists does element-wise addition — it actually **concatenates** the lists! NumPy arrays behave completely differently (and much more usefully for numeric work): `array1 + array2` adds element-wise.

### ML Connection
Broadcasting is what lets you apply a single set of learned weights or a single normalization formula across an entire dataset in one line, instead of writing a manual loop over every row.

<a id="vectorization"></a>
# 12. Vectorization vs Loops

"Vectorization" means expressing an operation as a whole-array computation instead of a manual Python loop. Vectorized NumPy code is both **shorter** and **much faster**. Let's see the speed difference directly.

```python
import time

size = 1_000_000
python_list = list(range(size))
numpy_array = np.arange(size)

# Pure Python loop
start = time.time()
result_list = [x * 2 for x in python_list]
python_time = time.time() - start

# Vectorized NumPy operation
start = time.time()
result_array = numpy_array * 2
numpy_time = time.time() - start

print(f"Python loop time: {python_time:.4f} seconds")
print(f"NumPy vectorized time: {numpy_time:.4f} seconds")
print(f"NumPy was roughly {python_time / numpy_time:.1f}x faster")
```

Exact numbers vary by machine, but NumPy is typically **dramatically** faster. This is precisely why ML libraries insist on NumPy (or NumPy-like) arrays instead of plain Python lists — training a model can involve billions of numeric operations.

<a id="aggregates"></a>
# 13. Aggregate Functions and the `axis` Parameter

NumPy provides fast aggregate functions: `sum`, `mean`, `std` (standard deviation), `min`, `max`, `argmin` (index of minimum), `argmax` (index of maximum). For 2D arrays, the `axis` parameter controls whether you aggregate down columns or across rows.

- `axis=0` → operate **down each column** (result has one value per column)
- `axis=1` → operate **across each row** (result has one value per row)

```python
student_marks = np.array([
    [78, 85, 90],   # Ali
    [88, 92, 95],   # Sara
    [55, 62, 58],   # Ahmed
])

print("Overall mean:", student_marks.mean())
print("Overall max:", student_marks.max())
print("Standard deviation:", student_marks.std())

print("Average per subject (axis=0):", student_marks.mean(axis=0))   # one value per column
print("Average per student (axis=1):", student_marks.mean(axis=1))   # one value per row

print("Index of top student (by average):", student_marks.mean(axis=1).argmax())
```

**Common mistake:** Mixing up `axis=0` and `axis=1`. A helpful trick: `axis=0` collapses the **rows** (moving *down* the columns), and `axis=1` collapses the **columns** (moving *across* each row).

### ML Connection
Computing a per-feature mean/standard deviation (`axis=0`) is exactly how you'd normalize each feature column in a dataset before training a model.

<a id="sorting-unique"></a>
# 14. Sorting and Unique Values

```python
marks = np.array([78, 85, 91, 67, 88])

print(np.sort(marks))              # ascending sorted copy (original untouched)
print(marks)                       # confirm original is unchanged
print(np.argsort(marks))           # indices that WOULD sort the array
```

```python
grades = np.array(["B", "A", "A", "C", "B", "A"])

print(np.unique(grades))                       # unique values, sorted
print(np.unique(grades, return_counts=True))   # unique values + how many times each appears
```

<a id="combining"></a>
# 15. Combining Arrays

- `np.concatenate` — join arrays along an existing axis.
- `np.vstack` — stack arrays vertically (as new rows).
- `np.hstack` — stack arrays horizontally (as new columns, for 1D: side-by-side).

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate([a, b]))    # [1 2 3 4 5 6]

row1 = np.array([[1, 2, 3]])
row2 = np.array([[4, 5, 6]])
print(np.vstack([row1, row2]))    # stacked as two rows

print(np.hstack([a, b]))          # same as concatenate here (1D arrays)
```

<a id="copy-view"></a>
# 16. Copy vs View

Slicing a NumPy array often returns a **view** — it shares the same underlying memory as the original! Modifying a view modifies the original array too. Use `.copy()` when you need a fully independent array.

```python
original = np.array([1, 2, 3, 4, 5])

view = original[1:4]        # this is a VIEW, not a copy
view[0] = 999                # modifying the view...
print(original)              # ...also changes the original array!
```

```python
original = np.array([1, 2, 3, 4, 5])

safe_copy = original[1:4].copy()   # an explicit, independent copy
safe_copy[0] = 999
print(original)     # original is unaffected
print(safe_copy)
```

**Common mistake:** Assuming a slice is always a safe, independent copy (like it is for Python lists). In NumPy, this is one of the most common sources of confusing bugs — when in doubt, call `.copy()` explicitly.

# 17. Basic Linear Algebra

Machine Learning is built on linear algebra. Two operations you'll see constantly:
- **Dot product** (`np.dot` or `@`): multiplies corresponding elements and sums the result — used to combine features with weights.
- **Matrix multiplication** (`@`): the 2D generalization of the dot product.

```python
features = np.array([2, 3, 1])     # e.g. [hours_studied, previous_score, attendance]
weights = np.array([0.5, 0.3, 0.2])   # a toy set of "learned" weights

prediction = np.dot(features, weights)   # equivalent to: (2*0.5)+(3*0.3)+(1*0.2)
print(prediction)

# The '@' operator is a shorthand for the same operation
print(features @ weights)
```

```python
# Matrix multiplication: applying the same weights to MULTIPLE samples at once
X = np.array([
    [2, 3, 1],   # sample 1
    [5, 1, 4],   # sample 2
    [1, 1, 1],   # sample 3
])
weights = np.array([0.5, 0.3, 0.2])

predictions = X @ weights   # one prediction per row/sample
print(predictions)
```

### ML Connection
`X @ weights` above is *literally* how a linear regression model computes its predictions for many samples at once — this single line replaces what would otherwise be a nested Python loop over every sample and every feature.

<a id="saving-loading"></a>
# 18. Saving and Loading Arrays

Just like the serialization concepts from Core Python (`json`, `pickle`), NumPy has its own fast, efficient binary format for saving arrays to disk: `.npy` for a single array, and `.npz` for multiple arrays bundled together.

```python
import os

os.makedirs("sample_data", exist_ok=True)

marks = np.array([78, 85, 91, 67, 88])

np.save("sample_data/marks.npy", marks)     # save a single array
loaded_marks = np.load("sample_data/marks.npy")   # load it back
print(loaded_marks)
print(np.array_equal(marks, loaded_marks))   # confirm the data matches exactly
```

```python
# Saving multiple arrays together in one .npz file
features = np.array([[2, 3, 1], [5, 1, 4]])
labels = np.array([1, 0])

np.savez("sample_data/dataset.npz", features=features, labels=labels)

loaded = np.load("sample_data/dataset.npz")
print(loaded["features"])
print(loaded["labels"])
```

### ML Connection
> Saving prepared feature arrays (and their labels) to `.npy`/`.npz` files means you don't have to reload and reprocess raw data every time you experiment with a model — directly building on the serialization ideas from the Core Python notebooks.

# 19. Common Mistakes — Recap

| Mistake | Fix |
|---|---|
| Mixing data types inside `np.array([...])` | Keep arrays single-typed for numeric work |
| Confusing `arange(start, stop, step)` with `linspace(start, stop, num)` | `arange` controls step size; `linspace` controls point count |
| Using `array1[1][2]` instead of `array1[1, 2]` for 2D indexing | Prefer the comma form — it's the NumPy idiom and faster |
| Reshaping into a shape whose product doesn't match the array's size | The product of the new shape must equal `.size` |
| Confusing `axis=0` vs `axis=1` | `axis=0` collapses down columns; `axis=1` collapses across rows |
| Assuming a slice is an independent copy | NumPy slices are often **views** — use `.copy()` when needed |
| Using `list1 + list2` and expecting element-wise addition | Only NumPy arrays add element-wise; Python lists concatenate |

# Summary

- NumPy's `ndarray` is a fast, memory-efficient, single-typed array — the numeric backbone of the entire ML/Data Science stack.
- You can create arrays from lists, or with generator functions (`zeros`, `ones`, `arange`, `linspace`, random generators).
- `.shape`, `.ndim`, `.size`, and `.dtype` describe an array's structure.

- Vectorized operations are dramatically faster than equivalent Python loops.
- Aggregate functions with the `axis` parameter let you summarize data per row or per column.
- Slices can be views (sharing memory) rather than independent copies — use `.copy()` when you need safety.
- Basic linear algebra (`@`, `np.dot`) is the literal mechanism behind how ML models compute predictions.
- `.npy`/`.npz` files let you save and reload prepared numeric data efficiently.
