````python
# Task 9 -- Your code here
df_copy = df.copy()
df_copy.loc[0, "salary"] = np.nan
print(df_copy["salary"].isna().sum())
average_salary = df_copy["salary"].mean()
df_copy["salary"].fillna(average_salary, inplace=True)
print(df_copy["salary"])
````

````python
# Task 10 -- Your code here

(df_copy.sort_values(by="department", ascending=False, inplace=True))
print(df_copy)
print(df_copy.sort_values(by="department", ascending=True, inplace=True))
print(df_copy)
````

### Task 11: Feature Engineering

### Task 12: Cleaning a Messier Dataset
Create a new DataFrame from this dictionary:
```python
messy = {
    "name": ["Ali", "Sara", None, "Ayesha", "Bilal"],
    "age": [28, np.nan, 45, 24, 38],
    "salary": [85000, 72000, 120000, np.nan, 98000],
}
```
1. Print how many missing values are in each column.
2. Drop any row where `name` is missing (hint: `df.dropna(subset=["name"])`).

4. Print the final, cleaned DataFrame.

### Task 13: A Small Exploration Report
Using the original `df` (employees dataset), write code that prints, in a clean format:
- The total number of employees.
- The average salary, rounded to 2 decimal places.
- The department with the highest average salary (hint: you'll need to compute an average salary per department — a preview of `groupby`, which you'll learn properly in Part 2. For now, try filtering + `.mean()` for each department one at a time, or explore if `df.groupby("department")["salary"].mean()` works for you).
- The name of the single highest-paid employee.

````python
# Task 11 -- Your code here
df_copy["Senior_Employees"] = ["True" if age > 40 else "False" for age in df_copy["age"]]
print(df_copy)
mask = df_copy["Senior_Employees"] == "True"
print(df_copy[mask])
````

````python
# Task 12 -- Your code here
messy = {
    "name": ["Ali", "Sara", None, "Ayesha", "Bilal"],
    "age": [28, np.nan, 45, 24, 38],
    "salary": [85000, 72000, 120000, np.nan, 98000],
}
df_messy = pd.DataFrame(messy)
print(df_messy["name"].isna().sum())
print(df_messy["age"].isna().sum())
print(df_messy["salary"].isna().sum())
df_messy = df_messy.dropna(subset=["name"])
df_messy["age"] = df_messy["age"].fillna(df_messy["age"].mean() , inplace=True)
df_messy["salary"] = df_messy["salary"].fillna(df_messy["salary"].mean() , inplace=True)
print(df_messy)
````

````python
# Task 13 -- Your code here
df = pd.read_csv("sample_data/employees.csv")
print(df["name"].count())
print(round(df['salary'].mean() , 2))
df.groupby("department")["salary"].mean()
````
