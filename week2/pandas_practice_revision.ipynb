Task 7: Missing Value Imputation
Create the messy DataFrame below:
df_messy = pd.DataFrame({
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "age": [22, np.nan, 23, 20],
    "marks": [84, 91, np.nan, 90]
})

•	1. Print the count of missing values per column using .isna().sum().
•	2. Fill missing values in "age" with the mean of "age", and missing values in "marks" with the mean of "marks" in a single .fillna() call using a dictionary.


```python
import numpy as np
import pandas as pd
df_messy = pd.DataFrame({
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "age": [22, np.nan, 23, 20],
    "marks": [84, 91, np.nan, 90]
})
print(df_messy.isna().sum())
df_messy['age']=df_messy['age'].fillna(df_messy['age'].mean().round(2), inplace=True)
df_messy['marks']=df_messy['marks'].fillna(df_messy['marks'].mean().round(2), inplace=True)
print(df_messy)
```

Task 8: String Cleaning with the .str Accessor
Given:
df_dept = pd.DataFrame({

Task 9: Relational Merge and Missing Keys
Given two DataFrames:
employees = pd.DataFrame({
    "emp_id": [101, 102, 103, 104],
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "dept_id": ["D1", "D2", "D1", "D99"]
})

departments = pd.DataFrame({
    "dept_id": ["D1", "D2", "D3"],
    "dept_name": ["Engineering", "Marketing", "Finance"]
})

Merge employees and departments on "dept_id" using a left join so that no employee rows are lost. Print the result and note how missing department matches are populated.


```python
employees = pd.DataFrame({
    "emp_id": [101, 102, 103, 104],
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "dept_id": ["D1", "D2", "D1", "D99"]
})
departments = pd.DataFrame({
    "dept_id": ["D1", "D2", "D3"],
    "dept_name": ["Engineering", "Marketing", "Finance"]
})
new_df = pd.merge(employees, departments, on='dept_id', how='left')
print(new_df)
```

Task 10: Concatenation and Deduplication
Given two batches:
batch_1 = pd.DataFrame({"item": ["A", "B", "C"], "price": [10, 20, 30]})
batch_2 = pd.DataFrame({"item": ["B", "D", "E"], "price": [20, 40, 50]})

•	1. Stack the two DataFrames vertically using pd.concat(), passing ignore_index=True to prevent duplicate index labels.
•	2. Identify duplicate items with .duplicated(subset=["item"]) and drop duplicates keeping only the first occurrence via .drop_duplicates().


```python
batch_1 = pd.DataFrame({"item": ["A", "B", "C"], "price": [10, 20, 30]})
batch_2 = pd.DataFrame({"item": ["B", "D", "E"], "price": [20, 40, 50]})

new_batch = pd.concat([batch_1,batch_2], ignore_index=True)
print(new_batch)
print(new_batch.duplicated(subset=["item"]))
print(new_batch.drop_duplicates())
```
